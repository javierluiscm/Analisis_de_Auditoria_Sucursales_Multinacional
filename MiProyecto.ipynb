## **Fase 2: Valores Atipicos**

Identificar y tratar los valores atípicos usando una técnica ya sea el método de Z-Score o el Rango intercuartílico para la variable numérica (audit_score ), limpiar los datos si es necesario y, finalmente, evaluar la correlación entre variables.

## 2.1 Identificar valores atípicos

Método 1: Z-Score

```python
import numpy as np

# Calcular Z-Score para audit_score
z_scores = (df['audit_score'] - df['audit_score'].mean()) / df['audit_score'].std()

# Filtrar valores atípicos (|z| > 3)
outliers_z = df[z_scores.abs() > 3]
print("Valores atípicos con Z-Score:")
print(outliers_z)
```

## Conclusión:

* Se detectaron registros con valores extremos de audit_score, como 1.0 y 100.0, que se salen de lo esperado.

* También aparecen valores muy altos (95–100) y muy bajos (≈1–2), que el método estadístico marca como outliers.

* Estos casos representan auditorías muy diferentes al comportamiento típico, ya sea con puntajes mínimos o máximos.

Rango Intercuartílico (IQR)

```python
Q1 = df['audit_score'].quantile(0.25)
Q3 = df['audit_score'].quantile(0.75)
IQR = Q3 - Q1

# Definir límites
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Filtrar valores atípicos
outliers_iqr = df[(df['audit_score'] < limite_inferior) | (df['audit_score'] > limite_superior)]
print("Valores atípicos con IQR:")
print(outliers_iqr)
```

## Conclusión:

* Se identificaron 77 registros como valores atípicos.

* Estos casos incluyen auditorías con puntajes muy bajos (≈1–4) y muy altos (≈91–100).

* También aparecen auditorías con duraciones inusuales (ejemplo: más de 20 días), que se salen del rango típico definido por el IQR.

## 2.2 Tratar los valores atípicos

```python
# 1. Calcular IQR
Q1 = df['audit_score'].quantile(0.25)
Q3 = df['audit_score'].quantile(0.75)
IQR = Q3 - Q1

# 2. Definir límites
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# 3. Calcular la mediana (para reemplazar outliers)
mediana = df['audit_score'].median()

# 4. Reemplazar valores atípicos por la mediana
df['audit_score_normalizado'] = df['audit_score'].apply(
    lambda x: mediana if (x < limite_inferior or x > limite_superior) else x
)

# 5. Verificar resultados
print("Mediana usada para reemplazo:", mediana)
print(df[['audit_score','audit_score_normalizado']].head(20))
```

## Conclusión:

Los valores extremos fueron suavizados al nivel de la mediana (49.9932...) . Esto hace que el conjunto de datos sea más estable y representativo del comportamiento típico.

## **3. FASE 3: Inferencia estadística:**  

Aquí el estudiante deberá realizar un muestreo aleatorio simple de los datos, antes de ellos tuvo que haber seleccionado el tamaño de la muestra. Luego, se podría plantear la siguiente prueba de hipótesis:

* H0: Las sucursales que cumplen con las normas (compliance=1) tienen la misma audit_score promedio que las que no cumplen (compliance=0).
* H1: Las sucursales que cumplen con las normas (compliance=1) tienen una audit_score promedio diferente a las que no cumplen (compliance=0).

El estudiante podría usar una prueba t de Student para evaluar esta hipótesis.


## Seleccionar el tamaño de la muestra

```python
import math

# Parámetros básicos
N = len(df)  # tamaño de la población (número total de registros)
Z = 1.96     # valor Z para 95% de confianza
p = 0.5      # proporción esperada (máxima variabilidad)
q = 1 - p
e = 0.05     # margen de error (5%)

# Fórmula para tamaño de muestra
n = (Z**2 * p * q) / (e**2)

# Ajuste para población finita
n_ajustado = n / (1 + (n - 1)/N)

print("Tamaño de muestra recomendado:", math.ceil(n_ajustado))
```

## Conclusión:

* Esto indica que el tamaño mínimo de muestra es 370 registros para garantizar representatividad con 95% de confianza y 5% de error.

* Cada registro tuvo la misma probabilidad de ser elegido, por lo que la muestra es imparcial y representativa.

## Prueba de hipótesis con t de Student:

* H0: Las sucursales que cumplen con las normas (compliance=1) tienen la misma audit_score promedio que las que no cumplen (compliance=0).

* H1: Las sucursales que cumplen con las normas (compliance=1) tienen una audit_score promedio diferente a las que no cumplen (compliance=0).

```python
from scipy.stats import ttest_ind

# Separar los grupos
grupo_cumple = df[df['compliance'] == 1]['audit_score_normalizado']
grupo_no_cumple = df[df['compliance'] == 0]['audit_score_normalizado']

# Prueba t de Student (asumiendo varianzas iguales)
t_stat, p_value = ttest_ind(grupo_cumple, grupo_no_cumple, equal_var=True)

print("Estadístico t:", t_stat)
print("p-value:", p_value)

# Interpretación:

if p_value > 0.05:
    print("No se rechaza H0: Los promedios son iguales.")
else:
    print("Se rechaza H0: Los promedios son diferentes.")
```

## Conclusión:

 * Estadístico t = 1.287 → indica la diferencia entre los promedios de los dos grupos.

* p-value = 0.198 → este valor es mayor a 0.05, lo que significa que la diferencia no es estadísticamente significativa.

**Conclusión: No se rechaza la hipótesis nula (H0).**

La prueba t muestra que no existe diferencia significativa en el puntaje promedio de auditoría entre sucursales cumplidoras y no cumplidoras.
Esto sugiere que el audit_score depende de otros factores (como severidad o duración), más que del cumplimiento directo.

**Pruebas de normalidad (Shapiro-Wilk y D’Agostino):**

**En conclusión la variable ES PARAMÉTRICA**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import statsmodels.api as sm
```

```python
## Definimos la variable.

# Seleccionamos la columna audit_score_normalizado
data = df['audit_score_normalizado']
```

```python
## Ajustar distribuciones comunes

# Ajuste Normal
params_norm = stats.norm.fit(data) ## Calcula los parámetros de la distribución Normal (media,desviación estándar)
loglik_norm = np.sum(stats.norm.logpdf(data, *params_norm)) ## Calcula el Log-Likelihood: qué tan probable es los datos provengan de esa distribución Normal.
aic_norm = 2*2 - 2*loglik_norm   ## Calcula los criterios de ajuste
                                 # 2 parámetros: media y desviación
bic_norm = np.log(len(data))*2 - 2*loglik_norm

# Ajuste Exponencial
params_exp = stats.expon.fit(data) ## Ajusta la distribución Exponencial (parámetro principal: tasa λ)
loglik_exp = np.sum(stats.expon.logpdf(data, *params_exp))
aic_exp = 2*2 - 2*loglik_exp       ## Calcula AIC y BIC para la Exponencial (usa 2 parámetros: ubicación y escala)
bic_exp = np.log(len(data))*2 - 2*loglik_exp

# Ajuste Gamma
params_gamma = stats.gamma.fit(data) ## Ajusta la distribución Gamma (3 parámetros: forma, ubicación y escala).
loglik_gamma = np.sum(stats.gamma.logpdf(data, *params_gamma))
aic_gamma = 2*3 - 2*loglik_gamma   # 3 parámetros: Calcula AIC y BIC para la Gamma (usa 3 parámetros).
bic_gamma = np.log(len(data))*3 - 2*loglik_gamma
```

## Calcula los criterios de ajuste:

* AIC (Akaike Information Criterion) → penaliza el número de parámetros, busca el modelo más simple y mejor ajustado.

* BIC (Bayesian Information Criterion) → similar al AIC, pero penaliza más cuando hay muchos datos.

```python
## Comparamos Criterios:

resultados = pd.DataFrame({
    "Distribución": ["Normal","Exponencial","Gamma"],
    "AIC": [aic_norm, aic_exp, aic_gamma],
    "BIC": [bic_norm, bic_exp, bic_gamma],
    "LogLik": [loglik_norm, loglik_exp, loglik_gamma]
})

print(resultados)
```

## Conclusión:

* El criterio dice que mientras más bajo el AIC y BIC, mejor el ajuste.

* La Normal tiene los valores más bajos → es la mejor distribución los datos.

* La Gamma también se ajusta bien, pero no supera a la Normal.

* La Exponencial se ajusta mal (AIC y BIC muy altos).

**Esto confirma que la variable es paramétrica y se comporta como una distribución normal.**
